```python
def descomposicion_LU_doolittle(A):
    """
    La funcion descompone una matriz A en el producto L × U, usando el metodo de Doolittle.
    A: matriz cuadrada 
    Retorna: L (triangular inferior), U (triangular superior)
    """
    n = len(A)
    L,  U= [], []

    # Verificar que sea matriz cuadrada
    for fila in A:
        if len(fila) != n:
            print("No es una matriz cuadrada")
            return None, None
    
    # Llenar matrices LU
    for i in range(n):
        fila_L = []
        fila_U = []
        for j in range(n):
            if i == j:
                fila_L.append(1)
            else:
                fila_L.append(0)
            fila_U.append(0)
        L.append(fila_L)
        U.append(fila_U)

    # Calculo de LU
    for i in range(n):
        for k in range(i, n):
            suma = 0
            for j in range(i):
                suma += L[i][j] * U[j][k]
            U[i][k] = A[i][k] - suma
    
        for k in range(i+1, n):
            suma = 0
            for j in range(i):
                suma += L[k][j] * U[j][i]
            L[k][i] = (A[k][i] - suma) / U[i][i]
    
    # Impresion de L
    print("L = ")
    for row in range(n):
        for col in range(n):
            print(f"{L[row][col]:.6f}", end=" ")
        print("")

    # Impresion de U
    print("\n U = ")
    for row in range(n):
        for col in range(n):
            print(f"{U[row][col]:.6f}", end=" ")
        print("")
    
    return L, U
```

```python
import math

def metodo_cholesky(A):
    n = len(A)
    L = []
    
    # Verificar que la matriz sea cuadrada
    for fila in A:
        if len(fila) != n:
            print("No es una matriz cuadrada")
            return None, None

    # Verificar que la matriz sea simetrica
    for i in range(n):
        for j in range(n):
            if A[i][j] != A[j][i]:
                print("La matriz no es simetrica")
                return None, None

    # Llenar matriz L con 0
    for i in range(n):
        fila = []
        for j in range(n):
            fila.append(0)
        L.append(fila)

    # Calculo de L
    for i in range(n):
        suma = 0 
        for k in range(i):
            suma += L[i][k] * L[i][k]
        L[i][i] = math.sqrt(A[i][i] - suma)
        for j in range(i+1, n):
            suma = 0
            for k in range(i):
                suma += L[j][k] * L[i][k]
            L[j][i] = (A[j][i] - suma) / L[i][i]    

    # Calculo de la matriz transpuesta de L
    L_T = []
    for i in range(n):
        fila_T = []
        for j in range(n):
            fila_T.append(L[j][i])
        L_T.append(fila_T)

    # Imprimir L
    print("L = ")
    for row in range(n):
        for col in range(n):
            print(f"{L[row][col]:.6f}", end=" ")
        print("")

    # Imprimir L_T
    print("\nL_T = ")
    for row in range(n):
        for col in range(n):
            print(f"{L_T[row][col]:.6f}", end=" ")
        print("")
    
    return L, L_T

```

```python
def gauss_jacobi(A, b, x0, tol, max_iter):
    """
    A: matriz de coeficientes (lista de listas)
    b: vector de términos independientes (lista)
    x0: vector inicial (lista)
    tol: tolerancia para la convergencia
    max_iter: número máximo de iteraciones
    Retorna: vector solución x
    """
    n = len(A)
    xO = x0.copy()
    x = [0] * n
    k = 1
    while k <= max_iter:
        print(f"Iteración {k}:")
        for i in range(n):
                suma = 0
                for j in range(n):
                    if i != j:
                        suma += A[i][j] * xO[j]
                x[i] = (b[i] - suma) / A[i][i]
                print(f"x{i+1} = {x[i]:.6f}")
    
        # Verificar tolerancia
        error_abs = []
        for i in range(n):
            error_abs.append(abs(x[i] - xO[i]))
        error_abs = max(error_abs)
        print(f"Error absoluto: {error_abs:.6f}")
        if error_abs < tol:
            print(f"\n El procedimiento fue exitoso con {k} iteraciones y un error abs de {error_abs} \n Soluciones:")
            for i in range(n):
                print(f"x{i+1} = {x[i]:.6f}")
            return x

        xO = x.copy() 
        k += 1
        
    print("Número máximo de iteraciones excedido")
    return x
```

```python
def gauss_seidel(A, b, x0, tol, max_iter):
    n = len(A)
    xO = x0.copy()
    x = [0] * n
    k = 1

    while k <= max_iter:
        print(f"Interación {k}:")
        for i in range(n):
            suma = 0
            for j in range(i):
                suma += A[i][j] * x[j]
            for j in range(i+1, n):
                suma += A[i][j] * xO[j]
            x[i] = (b[i] - suma) / A[i][i]
            print(f"x{i+1} = {x[i]:.6f}")  
        
        # Verificar tolerancia
        error_abs = []
        for i in range(n):
            error_abs.append(abs(x[i] - xO[i])) 
        error_abs = max(error_abs)
        print(f"Error absoluto: {error_abs:.6f}")
        if error_abs < tol:
            print(f"\nEl procedimiento fue exitoso con {k} iteraciones y un error abs de {error_abs} \n Soluciones:")
            for i in range(n):
                print(f"x{i+1} = {x[i]:.6f}")
            return x
        xO = x.copy()
        k += 1
    
    print("Número máximo de iteraciones excedido")
    return x
```
