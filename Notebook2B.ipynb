# Eliminación Gaussiana

```python
def eliminacion_gaussiana(matriz_A):
    n = len(matriz_A)

    # Hacer ceros hacia abajo
    for col in range(n-1):
        for row in range(col+1, n):
            factor = matriz_A[row][col]/matriz_A[col][col]
            for j in range(col, n+1):
                matriz_A[row][j] -= factor * matriz_A[col][j]
        if matriz_A[col][col] == 0:
            print("No existe solucion unica")
            return
            
    x=[0]*n
    for i in range(n-1, -1, -1):
        suma = 0
        for j in range(i+1, n):
            suma += matriz_A[i][j] * x[j]
        x[i] = (matriz_A[i][n] - suma) / matriz_A[i][i]
        
    # Impresión de la solución
    for i in range(n):
        print(f"x{i+1} = {x[i]:.6f}")


```

```python
matriz_aumentada = [    [3, 2, 1, 1, 1],
    [2, 3, 3, 2, 4],
    [5, 4, 6, 3, 7],
    [4, 3, 5, 6, 10] ]

print(eliminacion_gaussiana(matriz_aumentada))
```

# Gauss-Jordan

```python
def gauss_jordan(matriz_A):
    n = len(matriz_A)

    # Hacer ceros hacia abajo
    for col in range(n-1):
        for row in range(col+1, n):
            factor = matriz_A[row][col]/matriz_A[col][col]
            for j in range(col, n+1):
                matriz_A[row][j] -= factor * matriz_A[col][j]
        if matriz_A[col][col] == 0:
            print("No existe solucion unica")
            return    
        
    # Hacer ceros hacia arriba
    for col in range(n-1, 0, -1):
        for row in range(col-1, -1, -1):
            factor = matriz_A[row][col]/ matriz_A[col][col]
            for j in range(n+1):
                matriz_A[row][j] -= factor * matriz_A[col][j]

    # Hacer 1 en la diagonal
    for i in range(n):
        valor = matriz_A[i][i]
        for j in range(n+1):
            matriz_A[i][j] /= valor

    # Soluciones
    for i in range(n):
        print(f"x{i+1} = {matriz_A[i][n]:.6f}")
```

## Método de la Matriz Inversa

```python
def determinante(matriz):
    n = len(matriz)
    deter = 1
    A = [row[:] for row in matriz]
    
    # Verificar que la matriz sea cuadrada
    for row in matriz:
        if len(row) != n:
            print("No es una matriz cuadrada")
            return None

    if n == 1:
        return matriz[0][0]
    if n == 2:
        return matriz[0][0] * matriz[1][1] - matriz[0][1] * matriz[1][0]
    for k in range(n):
        max_row = k
        for i in range(k+1, n):
            if abs(A[i][k]) > abs(A[max_row][k]):
                max_row = i 
        if A[max_row][k] == 0:
            return 0
        if max_row != k:
            A[k], A[max_row] = A[max_row], A[k]
            deter = -deter
        deter *= A[k][k]
        for row in range(k+1, n):
            factor = A[row][k] / A[k][k]
            for col in range(k, n):
                A[row][col] -= factor * A[k][col]

    if deter == 0:
        print("El sistema no tiene solucion unica, el determinante es 0")
    else:
        print("El sistema tiene solucion unica y el determinante es:", round(deter, 2))
```

```python
def metodo_matriz_inversa(A, b):
    """
    A: matriz de coeficientes (lista de listas)
    b: vector de términos independientes (lista)
    Resuelve Ax = b calculando x = A⁻¹ * b
    """
    n = len(A)
    
    # Matriz indentidad
    identidad = []
    for i in range(n):
        fila = [0]*n
        fila[i] = 1
        identidad.append(fila)

    # Matriz aumentada [A | I]
    matriz_aumentada = []
    for i in range(n):
        fila = A[i] + identidad[i]
        matriz_aumentada.append(fila)

    # Calcular la matriz inversa
    for col in range(n):
        for row in range(col+1, n):
            factor = matriz_aumentada[row][col] / matriz_aumentada[col][col]
            for j in range(2*n):
                matriz_aumentada[row][j] -= factor * matriz_aumentada[col][j]
    
    for col in range(n-1, -1, -1):
        for row in range(col-1, -1, -1):
            factor = matriz_aumentada[row][col] / matriz_aumentada[col][col]
            for j in range(2*n):
                matriz_aumentada[row][j] -= factor * matriz_aumentada[col][j]
    
    for i in range(n):
        valor = matriz_aumentada[i][i]
        for j in range(2*n):
            matriz_aumentada[i][j] /= valor

    # Extraer la matriz inversa
    A_inversa = []
    for i in range(n):
        fila_inversa = []
        for j in range(n, 2*n):
            fila_inversa.append(matriz_aumentada[i][j])
        A_inversa.append(fila_inversa)

    # Imprimir la matriz inversa
    print("Matriz Inversa A⁻¹:")
    for row in range(n):
        for col in range(n):
            print(f"{A_inversa[row][col]:.6f}", end=" ")
        print()

    # Multiplicar A⁻¹ * b para obtener los valores de x
    x_valores = []
    for i in range(n):
        suma = 0
        for j in range(n):
            suma += A_inversa[i][j] * b[j]
        x_valores.append(suma)

    # Soluciones de x
    print("\n Soluciones:")
    for i in range(n):
        print(f"x{i+1} = {x_valores[i]:.6f}")
```

```python
A = [[1, 2, 3], [0, 1, 4], [5, 6, 0]]
b = [1, 2, 3]  
print("Verificar el determinante de la matriz A")
determinante(A)
print("\n")
metodo_matriz_inversa(A, b)
```

## Producto de Matrices, Doolittle

```python
def descomposicion_LU_doolittle(A):
    """
    La funcion descompone una matriz A en el producto L × U, usando el metodo de Doolittle.
    A: matriz cuadrada 
    Retorna: L (triangular inferior), U (triangular superior)
    """
    n = len(A)
    L,  U= [], []

    # Verificar que sea matriz cuadrada
    for fila in A:
        if len(fila) != n:
            print("No es una matriz cuadrada")
            return None, None
    
    # Llenar matrices LU
    for i in range(n):
        fila_L = []
        fila_U = []
        for j in range(n):
            if i == j:
                fila_L.append(1)
            else:
                fila_L.append(0)
            fila_U.append(0)
        L.append(fila_L)
        U.append(fila_U)

    # Calculo de LU
    for i in range(n):
        for k in range(i, n):
            suma = 0
            for j in range(i):
                suma += L[i][j] * U[j][k]
            U[i][k] = A[i][k] - suma
    
        for k in range(i+1, n):
            suma = 0
            for j in range(i):
                suma += L[k][j] * U[j][i]
            L[k][i] = (A[k][i] - suma) / U[i][i]
    
    # Impresion de L
    print("L = ")
    for row in range(n):
        for col in range(n):
            print(f"{L[row][col]:.6f}", end=" ")
        print("")

    # Impresion de U
    print("\n U = ")
    for row in range(n):
        for col in range(n):
            print(f"{U[row][col]:.6f}", end=" ")
        print("")
    
    return L, U
```

```python
matriz_A = [[2, -1, 1], [-3, 3, 9], [6, -6, -6]]
print(descomposicion_LU_doolittle(matriz_A))
```

## Método Crout

```python
def metodo_crout(A):
    n = len(A)
    L, U = [], []

    # Verificar que sea matriz cuadrada
    for fila in A:
        if len(fila) != n:
            print("No es una matriz cuadrada")
            return None, None
        
    # Llenar matrices LU
    for i in range(n):
        fila_L = []
        fila_U = []
        for j in range(n):
            if i == j:
                fila_U.append(1)
            else:
                fila_U.append(0)
            fila_L.append(0)
        L.append(fila_L)
        U.append(fila_U)

        # Calculo de LU
    for j in range(n):
        for i in range(j, n):
            suma = 0
            for k in range(j):
                suma += L[i][k] * U[k][j]
            L[i][j] = A[i][j] - suma

        for k in range(j+1, n):
            suma = 0
            for m in range(j):
                suma += L[j][m] * U[m][k]
            U[j][k] = (A[j][k] - suma) / L[j][j]

    # Impresion de L
    print("L = ")
    for row in range(n):
        for col in range(n):
            print(f"{L[row][col]:.6f}", end=" ")
        print("")

    # Impresion de U
    print("\n U = ")
    for row in range(n):
        for col in range(n):
            print(f"{U[row][col]:.6f}", end=" ")
        print("")

    return L, U
```

## Método Cholesky

```python
import math

def metodo_cholesky(A):
    n = len(A)
    L = []
    
    # Verificar que la matriz sea cuadrada
    for fila in A:
        if len(fila) != n:
            print("No es una matriz cuadrada")
            return None, None

    # Verificar que la matriz sea simetrica
    for i in range(n):
        for j in range(n):
            if A[i][j] != A[j][i]:
                print("La matriz no es simetrica")
                return None, None

    # Llenar matriz L con 0
    for i in range(n):
        fila = []
        for j in range(n):
            fila.append(0)
        L.append(fila)

    # Calculo de L
    for i in range(n):
        suma = 0 
        for k in range(i):
            suma += L[i][k] * L[i][k]
        L[i][i] = math.sqrt(A[i][i] - suma)
        for j in range(i+1, n):
            suma = 0
            for k in range(i):
                suma += L[j][k] * L[i][k]
            L[j][i] = (A[j][i] - suma) / L[i][i]    

    # Calculo de la matriz transpuesta de L
    L_T = []
    for i in range(n):
        fila_T = []
        for j in range(n):
            fila_T.append(L[j][i])
        L_T.append(fila_T)

    # Imprimir L
    print("L = ")
    for row in range(n):
        for col in range(n):
            print(f"{L[row][col]:.6f}", end=" ")
        print("")

    # Imprimir L_T
    print("\nL_T = ")
    for row in range(n):
        for col in range(n):
            print(f"{L_T[row][col]:.6f}", end=" ")
        print("")
    
    return L, L_T
```

```python
matriz_A = [[4, 2, 2], [2, 10, 4], [2, 4, 6]]

print(metodo_cholesky(matriz_A))
```

# Gauss Jacobi

```python
def gauss_jacobi(A, b, x0, tol, max_iter):
    """
    A: matriz de coeficientes (lista de listas)
    b: vector de términos independientes (lista)
    x0: vector inicial (lista)
    tol: tolerancia para la convergencia
    max_iter: número máximo de iteraciones
    Retorna: vector solución x
    """
    n = len(A)
    xO = x0.copy()
    x = [0] * n
    k = 1
    while k <= max_iter:
        print(f"Iteración {k}:")
        for i in range(n):
                suma = 0
                for j in range(n):
                    if i != j:
                        suma += A[i][j] * xO[j]
                x[i] = (b[i] - suma) / A[i][i]
                print(f"x{i+1} = {x[i]:.6f}")
    
        # Verificar tolerancia
        error_abs = []
        for i in range(n):
            error_abs.append(abs(x[i] - xO[i]))
        error_abs = max(error_abs)
        print(f"Error absoluto: {error_abs:.6f}")
        if error_abs < tol:
            print(f"\n El procedimiento fue exitoso con {k} iteraciones y un error abs de {error_abs} \n Soluciones:")
            for i in range(n):
                print(f"x{i+1} = {x[i]:.6f}")
            return x

        xO = x.copy() 
        k += 1
        
    print("Número máximo de iteraciones excedido")
    return x
```

```python
A = [[10, 1, 1], [2, 10, 1], [2, 2, 10]]
b = [12, 13, 14]
x0 = [0, 0, 0]
tol = 1e-4

gauss_jacobi(A, b, x0, tol, 100)
```

# Gauss Seidel

```python
def gauss_seidel(A, b, x0, tol, max_iter):
    n = len(A)
    xO = x0.copy()
    x = [0] * n
    k = 1

    while k <= max_iter:
        print(f"Interación {k}:")
        for i in range(n):
            suma = 0
            for j in range(i):
                suma += A[i][j] * x[j]
            for j in range(i+1, n):
                suma += A[i][j] * xO[j]
            x[i] = (b[i] - suma) / A[i][i]
            print(f"x{i+1} = {x[i]:.6f}")  
        
        # Verificar tolerancia
        error_abs = []
        for i in range(n):
            error_abs.append(abs(x[i] - xO[i])) 
        error_abs = max(error_abs)
        print(f"Error absoluto: {error_abs:.6f}")
        if error_abs < tol:
            print(f"\nEl procedimiento fue exitoso con {k} iteraciones y un error abs de {error_abs} \n Soluciones:")
            for i in range(n):
                print(f"x{i+1} = {x[i]:.6f}")
            return x
        xO = x.copy()
        k += 1
    
    print("Número máximo de iteraciones excedido")
    return x
```

```python
A = [[10, 1, 1], [2, 10, 1], [2, 2, 10]]
b = [6, 5, -1]
x0 = [0, 0, 0]
tol = 1e-4

gauss_seidel(A, b, x0, tol, 100)
```

# Método de Euler

```python
def metodo_euler(f, x0, y0, h, n):
    """
    f: función que define la ecuación diferencial
    x0: valor inicial de x
    y0: valor inicial de y
    h: tamaño del paso
    n: número de pasos a realizar
    Retorna: lista de puntos (x, y) calculados
    """
    x_valores = [x0]
    y_valores = [y0]
    x_actual = x0
    y_actual = y0

    print(f"Punto inicial: x₀ = {x0:.6f}, y₀ = {y0:.6f}")
    print("Iteraciones:")
    
    for i in range(1, n+1):
        pendiente = f(x_actual, y_actual)

        x_nuevo = x_actual + h
        y_nuevo = y_actual + h * pendiente

        print(f"Iteración {i}:")
        print(f"  f({x_actual:.6f}, {y_actual:.6f}) = {pendiente:.6f}")
        print(f"  x{i} = {x_actual:.6f} + {h} = {x_nuevo:.6f}")
        print(f"  y{i} = {y_actual:.6f} + {h} * {pendiente:.6f} = {y_nuevo:.6f}")

        x_valores.append(x_nuevo)
        y_valores.append(y_nuevo)

        x_actual = x_nuevo
        y_actual = y_nuevo

    # Mostrar resultados
    print("i\tx_i\t\ty_i\t\tf(x_i, y_i)\ty_i+1")
    print("-" * 60)
    for i in range(len(x_valores)):
        y_siguiente = f"{y_valores[i+1]:.6f}" if i < len(y_valores)- 1 else "N/A"
        print(f"{i}\t{x_valores[i]:.6f}\t{y_valores[i]:.6f}\t{f(x_valores[i], y_valores[i]):.6f}\t{y_siguiente}")

    return x_valores, y_valores
```

```python
def f(x, y):
    return x + y

x0 = 0.0
y0 = 1.0    
h = 0.2
n = 6

metodo_euler(f, x0, y0, h, n)
```
