```python
from scipy.optimize import curve_fit

def minimos_cuadrados_func(x, y, func):
    """
    Ajusta una función arbitraria a los datos (x, y) usando mínimos cuadrados.

    Parámetros:
    x (list o array): valores de x
    y (list o array): valores de y
    func (callable): función modelo, por ejemplo lambda x, a, b: a*x + b

    Retorna:
    popt (tuple): parámetros óptimos ajustados por mínimos cuadrados
    """
    x = np.array(x)
    y = np.array(y)
    popt, _ = curve_fit(func, x, y)
    print(f"Parámetros óptimos: {popt}")
    return popt
```

```python
import numpy as np

# Datos de ejemplo
x = np.array([0, 6, 12, 18, 24])
y = np.array([20, 26, 30, 26, 20])

# Modelo lineal: y = a*x + b
modelo = lambda t, a0, a1: a0 + a1 * np.cos(np.pi * t / 12)

# Probar el método de mínimos cuadrados
minimos_cuadrados_func(x, y, modelo)
```
