```python
from scipy.optimize import curve_fit

def minimos_cuadrados_func(x, y, func):
    """
    Ajusta una función arbitraria a los datos (x, y) usando mínimos cuadrados.

    Parámetros:
    x (list o array): valores de x
    y (list o array): valores de y
    func (callable): función modelo, por ejemplo lambda x, a, b: a*x + b

    Retorna:
    popt (tuple): parámetros óptimos ajustados por mínimos cuadrados
    """
    x = np.array(x)
    y = np.array(y)
    popt, _ = curve_fit(func, x, y)
    print(f"Parámetros óptimos: {popt}")
    return popt
```

```python
def metodo_euler_mejorado(f, x0, y0, h, n):
    """
    Resuelve una ecuación diferencial ordinaria dy/dx = f(x,y) usando el método de Euler mejorado.

    Parámetros:
    f (function): Función que define la ecuación diferencial dy/dx = f(x,y).
    x0 (float): Valor inicial de la variable independiente x.
    y0 (float): Valor inicial de la variable dependiente y correspondiente a x0.
    h (float): Tamaño del paso para la integración numérica.
    n (int): Número de pasos a realizar en la integración.

    Descripción:
    - Aplica el método de Euler mejorado.
    - Usa dos evaluaciones de la pendiente para obtener mayor precisión que Euler simple.
    - Calcula k₁ = h * f(x_i, y_i) como predictor inicial.
    - Calcula k₂ = h * f(x_i + h, y_i + k₁) como corrector.
    - Usa la fórmula: y_{i+1} = y_i + (k₁ + k₂)/2 para mayor exactitud.
    - Imprime el punto inicial y cada iteración con valores de k₁ y k₂.
    - Muestra una tabla final con todos los valores de x e y calculados.
    - Es un método de segundo orden que proporciona mejores aproximaciones que Euler simple.

    Retorna:
    tuple: (x_valores, y_valores) donde ambas son listas con los valores de x e y
    calculados en cada paso de la integración, incluyendo el punto inicial.
    """
    x_valores = [x0]
    y_valores = [y0]
    x_actual = x0
    y_actual = y0

    print(f"Punto inicial: x₀ = {x0:.6f}, y₀ = {y0:.6f}")
    print("Iteraciones:")
    for i in range(1, n+1):
        k1 = h* f(x_actual, y_actual)
        k2 = h * f(x_actual + h, y_actual + k1)

        x_nuevo = x_actual + h
        y_nuevo = y_actual + (k1 + k2) / 2

        print(f"Iteración {i}:")
        print(f"  k₁ = {k1:.6f}")
        print(f"  k₂ = {k2:.6f}")
        print(f"  x{i} = {x_actual:.6f} + {h} = {x_nuevo:.6f}")
        print(f"  y{i} = {y_actual:.6f} + ({k1:.6f} + {k2:.6f})/2 = {y_nuevo:.6f}")

        x_valores.append(x_nuevo)
        y_valores.append(y_nuevo)

        x_actual = x_nuevo
        y_actual = y_nuevo
    
    # Mostrar resultados
    print("i\tx_i\t\ty_i")
    print("-" * 30)
    for i in range(len(x_valores)):
        print(f"{i}\t{x_valores[i]:.6f}\t{y_valores[i]:.6f}")

    return x_valores, y_valores
```

```python
# Definir la función del problema: y' = t*exp(3*t) - 2*y
def f(t, y):
    return t * np.exp(3 * t) - 2 * y

# Parámetros iniciales
t0 = 0
y0 = 0
h = 0.5
n = int((1 - t0) / h)  # Número de pasos para llegar a t=1

# Probar el método de Euler mejorado con estos datos
metodo_euler_mejorado(f, t0, y0, h, n)
```
