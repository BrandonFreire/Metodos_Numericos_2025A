# Eliminación Gaussiana

Se debe cumplir:

1. Se requiere 1 ecuación por cada variable desconocida.

## Método gráfico

Se ponen los puntos $x$ y $y$ en un plano cartesiano y se unen con una línea recta. El punto de intersección de las líneas es la solución del sistema.

El problema es que se puede tener una infinita cantidad de puntos que cumplan con las condiciones del sistema, por lo que no es un método práctico para sistemas con más de dos variables.

La eliminación gaussiana es más preciso y menos costoso hablando computacionalmente. Además, tiene menor error porque a diferencia de Gauss-Jordan se puede encontrar con números decimales que toca redondear.
