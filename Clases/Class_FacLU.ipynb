La matriz inversa es la matriz $A^{-1}$ tal que $A A^{-1} = I$, donde $I$ es la matriz identidad.

El determinante es un número asociado a una matriz cuadrada que indica si la matriz es invertible. Si el determinante es cero, la matriz no tiene inversa.4
    Cuando el determinante es distinto de cero, el sistema de ecuaciones tiene solución única.

Matriz singular:
No tiene inversa, su determinante es cero.

Si el determinante es cero, por LU no se puede resolver, fin.

## Gauss-Jacobi

Es un método iterativo

Existe problemas con los métodos iterativos, como en la convergencia.

Se necesita una estimación inicial, que sirve para iniciar el proceso iterativo y si la estimación inicial es muy cercana al valor original, el proceso convergerá más rápido y por ende abrá menos iteraciones.

Formula:


$x_i^{(k+1)} = \frac{1}{a_{ii}} \left( b_i - \sum_{\substack{j=1 \\ j \neq i}}^{n} a_{ij} x_j^{(k)} \right), \quad i = 1, 2, \ldots, n$

La fórmula dice que la siguiente estimación debe estar ligada a la estimación anterior, es decir, la estimación actual depende de las estimaciones anteriores.

Respecto a la convergencia, este método no siempre puede converger. También, si los valores o las aproximaciones que empieza a dar puede que se vayan a alejando y acercando

Una de las maneras de saber si el método converge es ver si la matriz es diagonalmente dominante, es decir, si el valor absoluto del elemento de la diagonal es mayor que la suma de los valores absolutos de los demás elementos de la fila.

-> En estos métodos no se pueden realizar cambios de fila

La condición de diagonalmente dominante es:

$|a_{ii}| > \sum_{\substack{j=1 \\ j \neq i}}^{n} |a_{ij}|, \quad i = 1, 2, \ldots, n$

Las entradas para el código son: 

- La estimación inicial 
- La matriz de coeficientes $A$
- El número de ecuaciones y valores desconocidos $n$

-> No está permitido el uso de librerías en este método, excepto la librería de sumatoria

La última iteración es precisa y exacta.

Este método es paralelo: x, y, z, w, etc. se pueden calcular al mismo tiempo.
