## EDO

Se necesita un rango de tiempo, siempre de menor a mayor.

$a <= t <= b$

- La función debe ser continua y la derivada parcial debe ser continua en esa región.

Diferencias entre EDOS y derivadas parciales:

- La EDO solo tiene una variable independiente, mientras que las derivadas parciales tienen más de una.
- Es más fácil resolver una EDO que una derivada parcial, porque la EDO tiene una sola variable independiente.

## Método de Euler

Son de orden 1, es decir, solo hay una derivada.

Hay un punto tangente, es decir, una pendiente, que es la derivada de la función.

Es un método iterativo.

A partir del punto tangente, se van calculando los siguientes puntos de solución.

### Puntos de malla

Todos los puntos son equidistantes, es decir, tienen la misma distancia entre ellos.

Las derivadas son valorses discretos, es decir, se calculan en puntos específicos.

Esto sirve para discretizar el intervalo de iteración, es decir, 

-> Discretizar es pasar un valor continuo a un valor discreto.

Se va a pasar la EDO de un valor continuo a un valor discreto, es decir, se va a pasar de un intervalo continuo a un intervalo discreto.

¿Como se van a calcular los puntos de malla?

- Se van a calcular por interpolacion.
- Se van a calcular desde puntos ya calculados.
