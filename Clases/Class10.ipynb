# **Mínimos cuadrados**

Para aproximar una función existen varios métodos de aproximación. Por ejemplo en los splines cúbicos se logra suavizar la función en comparación de otros métodos. 

Los **minimos cuadrados** son un método de aproximación que minimiza la suma de los cuadrados de los errores (residuos) entre los valores observados y los valores predichos por un modelo. Este método es ampliamente utilizado en regresión lineal y otros modelos estadísticos.

**¿Qué pasa con los puntos atípicos?**

Cuanto más grande es el error, más grande es el mínimo cuadrado. Por lo tanto, los puntos atípicos tienen un gran impacto en el resultado del modelo.

¿Cómo se minimiza?

Se minimiza derivando con la regla de la cadena. 

Dados los puntos ${(x_1, y_1), (x_2, y_2), \ldots, (x_n, y_n)}$, el objetivo es encontrar una función $f(x)$ que minimice la suma de los cuadrados de los errores:



Como se va a derivar, siempre se va a tomar las derivadas parciales tomando en cuenta $a_i$ y $b_i$ como variables independientes.

**¿Que pasa con el error respecto a los coeficientes que se van a encontrar en las derivadas parciales?**

${E = \sum_{i=1}^{n} [y_i - f(x_i)]^2}$

Los mínimos cuadrados se van a interpolar a cualqueir tipo de función. 

¿Por qué los errores son cuadrados?

Porque si fuese más grande que 2, se incrementa.

Porque al elevar al cuadrado se evita que los errores negativos se cancelen con los positivos, y también penaliza más fuertemente los errores grandes.
