- Condición de frontera natural: $S''_0(x_0) = S''_{n-1}(x_n) = 0$

- Condición de frontera de tipo Dirichlet: $S_0(x_0) = y_0$ y $S_{n-1}(x_n) = y_n$

Los splines cúbicos son una forma de interpolación que permite crear una función suave a partir de un conjunto de puntos. La elección entre frontera natural y frontera condicional depende del contexto y los requisitos específicos del problema. Los splines hacen que siempre la función pase por los puntos que se tiene.
