# Exactitud vs Precisión
**Exactitud:** (Cerca del valor real) es la medida de cuán cerca está una medición del valor verdadero. Se refiere a la proximidad de un resultado a su valor real o verdadero. En otras palabras, es una medida de cuán correcto es un resultado en comparación con el valor real.

**Precisión:** (valores consistentes) es la medida de cuán consistentes son las mediciones entre sí. Se refiere a la cercanía de un conjunto de resultados entre sí, independientemente de si están cerca del valor verdadero o no. En otras palabras, es una medida de cuán repetibles son los resultados.

# Tipos de errores
### **1. Error de corte o truncamiento:**

```python
def truncar(numero, decimales):
    """
    Trunca un número decimal a una cantidad específica de decimales.

    Parámetros:
    numero (float): El número que se desea truncar.
    decimales (int): La cantidad de decimales a la que se desea truncar el número.

    Retorna:
    float: El número truncado a la cantidad de decimales especificada.
    """
    factor = 10.0 ** decimales
    return int(numero * factor) / factor

# Ejemplo de uso:
pi = math.pi

# Aplicaición:
pi_truncado = truncar(pi, 3)
print("Valor dado: ", pi)
print("Valor truncado a 3 decimales:", pi_truncado)
```

# Calculo de error

<div align="center">

$Error_{real} = p - p^*$

</div>   

```python
def error_real(valor_real, valor_aproximado):
    """
    Calcula el error real entre un valor real y un valor aproximado.

    Parámetros:
    valor_real (float): El valor verdadero o exacto.
    valor_aproximado (float): El valor aproximado o estimado.

    Retorna:
    float: El error real calculado como la diferencia entre valor_real y valor_aproximado.
    """
    return valor_real - valor_aproximado

# Error real
calculo_error_real = error_real(pi, pi_truncado)

# Valor de pi
print(pi)

print("Error real:", calculo_error_real)
```

<div align="center">

$Error_{absoluto} = |p - p^*|$

</div>   


```python
def error_absoluto(error_real):
    """
    Calcula el error absoluto a partir del error real.

    Parámetros:
    error_real (float): El error real calculado como la diferencia entre el valor real y el valor aproximado.

    Retorna:
    float: El valor absoluto del error real (error absoluto).
    """
    return abs(error_real)

calculo_error_abs = error_absoluto(calculo_error_real)

print("Error absoluto:", calculo_error_abs)
```


<div align="center">

$Error_{relativo} = \frac{| p - p^* |}{| p |} $

</div>   


```python
def error_relativo(valor_real, valor_aproximado):
    """
    Calcula el error relativo entre un valor real y un valor aproximado.

    Parámetros:
    valor_real (float): El valor verdadero o exacto.
    valor_aproximado (float): El valor aproximado o estimado.

    Retorna:
    float: El error relativo calculado como el valor absoluto del error real dividido para el valor real.
    """
    return abs((valor_real - valor_aproximado) / valor_real)

calculo_error_relativo = error_relativo(pi, pi_truncado)

print("Error relativo:", calculo_error_relativo)
```

<div align="center">

$Error_{relativo} = \frac{| p - p^* |}{| p |} \cdot  {100 \%}$

</div>   


```python
def error_relativo_porcentual(error_relativo):
    """
    Calcula el error relativo porcentual a partir del error relativo.

    Parámetros:
    error_relativo (float): El error relativo calculado como el valor absoluto del error real dividido para el valor real.

    Retorna:
    float: El error porcentual (error relativo multiplicado por 100).
    """
    return error_relativo * 100

calculo_error_relativo_porcentual = error_relativo_porcentual(calculo_error_relativo)

print("Error porcentual:", calculo_error_relativo_porcentual, "%")
```

# Representación Numérica 32 y 64 bits

Pasos:

1. Convertir a binario
2. Escribir en notación científica
3. Seguir estándar IEEE 754 32 bits

```python
# Paso 1

def decimal_a_binario(numero, bits):
    """
    Convierte un número decimal a su representación binaria, considerando la cantidad de bits para la precisión de la parte fraccionaria.

    Parámetros:
    numero (float): El número decimal que se desea convertir a binario.
    bits (int): La cantidad de bits (32 o 64) que determina la precisión de la parte fraccionaria según el estándar IEEE 754.

    Descripción:
    - Separa la parte entera y la parte fraccionaria del número.
    - Convierte la parte entera a binario.
    - Convierte la parte fraccionaria a binario, limitando la cantidad de dígitos según los bits especificados (24 para 32 bits, 53 para 64 bits, menos los bits usados por la parte entera).
    - Une ambas partes para obtener la representación binaria completa.

    Retorna:
    str: Una cadena de texto con la representación binaria del número, incluyendo la parte entera y fraccionaria separadas por un punto.
    """
    #Separar la parte entera y parte decimal del numero
    parte_entera = int(numero)
    parte_fraccionaria = numero - parte_entera

    #Convertir la parte entera a binario
    binario_entero = ""
    while parte_entera > 0:
        binario_entero = str(parte_entera % 2) + binario_entero
        parte_entera //= 2

    #Convertir la parte fraccionaria a binario
    binario_decimal = ""
    if bits == 32:
        tamanio_decimal = 24 - len(binario_entero)
        while parte_fraccionaria > 0 and len(binario_decimal) < tamanio_decimal:  
            parte_fraccionaria *= 2
            if parte_fraccionaria >= 1:
                binario_decimal += "1"
                parte_fraccionaria -= 1
            else:
                binario_decimal += "0"
    elif bits == 64:
        tamanio_decimal = 53 - len(binario_entero)
        while parte_fraccionaria > 0 and len(binario_decimal) < tamanio_decimal:  
            parte_fraccionaria *= 2
            if parte_fraccionaria >= 1:
                binario_decimal += "1"
                parte_fraccionaria -= 1
            else:
                binario_decimal += "0"
        
    if binario_entero == "":
        binario_entero = "0"

    if binario_decimal:
        return f"{binario_entero}.{binario_decimal}" 
    else:
        return binario_entero
```

```python
# Paso 2

def notacion_cientifica(binario):
    """
    Convierte una representación binaria en cadena a notación científica binaria.

    Parámetros:
    binario (str): Cadena que representa el número binario, con parte entera y fraccionaria separadas por un punto.

    Descripción:
    - Si la parte entera contiene un '1', mueve el punto decimal a la izquierda hasta después del primer '1'.
    - Si la parte entera es '0', busca el primer '1' en la parte fraccionaria y mueve el punto decimal a la derecha.
    - Calcula el exponente correspondiente al desplazamiento del punto decimal.
    - Construye la mantisa en notación científica binaria.

    Retorna:
    tuple: Una tupla (mantisa, exponente), donde mantisa es la cadena en notación científica binaria y exponente es un entero.
    """
    # Separar la parte entera y la parte fraccionaria
    if '.' in binario:
        parte_entera, parte_fraccionaria = binario.split('.')
    else:
        parte_entera, parte_fraccionaria = binario, ''

    # Determinar la posición del primer '1' en la parte entera
    if parte_entera != '0':
        # Si hay parte entera, mover el punto a la izquierda
        exponente = len(parte_entera) - 1
        mantisa = parte_entera + parte_fraccionaria
        mantisa = mantisa[0] + '.' + mantisa[1:]
    else:
        # Si no hay parte entera, mover el punto a la derecha
        exponente = -1
        for i, bit in enumerate(parte_fraccionaria):
            if bit == '1':
                exponente = -(i + 1)
                break
        mantisa = '1.' + parte_fraccionaria[i + 1:]
    return mantisa, exponente
```

```python
# Paso 3

def signo(numero):
    """
    Determina el bit de signo para la representación IEEE 754 de un número.

    Parámetros:
    numero (float): El número decimal que se desea convertir a formato IEEE 754.

    Descripción:
    - Si el número es negativo, retorna 1 (indicando bit de signo negativo).
    - Si el número es positivo o cero, retorna 0 (indicando bit de signo positivo).

    Retorna:
    int: 1 si el número es negativo, 0 si es positivo o cero.
    """
    if numero < 0:
        return 1
    else:
        return 0
```

```python
def exponente(exponente, bits):
    """
    Calcula la representación binaria del exponente ajustado según el estándar IEEE 754 para 32 o 64 bits.

    Parámetros:
    exponente (int): El exponente obtenido de la notación científica binaria (puede ser positivo o negativo).
    bits (int): Número de bits del formato IEEE 754 (32 para simple precisión, 64 para doble precisión).

    Descripción:
    - Ajusta el exponente sumando el sesgo correspondiente (127 para 32 bits, 1023 para 64 bits).
    - Convierte el exponente ajustado a su representación binaria usando la función decimal_a_binario.

    Retorna:
    str: Cadena con la representación binaria del exponente ajustado según el formato IEEE 754.
    """
    binario = ""
    if bits == 32:
        exponente += 127
        binario = decimal_a_binario(exponente, 32)
    elif bits == 64:
        exponente += 1023
        binario = decimal_a_binario(exponente, 64)
    return binario
```

```python
def mantisa(numero, bits):
    """
    Calcula la mantisa para la representación IEEE 754 de un número en notación científica binaria.

    Parámetros:
    numero (str): Cadena que representa el número en notación científica binaria (por ejemplo, '1.00011001100110011001100').
    bits (int): Número de bits del formato IEEE 754 (32 para simple precisión, 64 para doble precisión).

    Descripción:
    - Separa la parte entera y fraccionaria de la cadena binaria.
    - Elimina el primer bit '1'.
    - Completa con ceros a la derecha hasta alcanzar la longitud requerida por el estándar (23 bits para 32 bits, 52 bits para 64 bits).
    - Recorta la mantisa si es necesario para ajustarse al tamaño.

    Retorna:
    str: Cadena binaria que representa la mantisa ajustada según el formato IEEE 754.
    """
    # Separar la parte entera y la parte fraccionaria
    if '.' in numero:
        parte_entera, parte_fraccionaria = numero.split('.')
    else:
        parte_entera, parte_fraccionaria = numero, ''

    # Obtener la mantisa
    mantisa = parte_entera + parte_fraccionaria
    mantisa = mantisa[1:]  # Eliminar el primer '1'

    if bits == 32:
        # Completar con ceros hasta 23 bits
        mantisa += '0' * (23 - len(mantisa))
        mantisa = mantisa[:23]

    elif bits == 64:
        # Completar con ceros hasta 52 bits
        mantisa += '0' * (52 - len(mantisa))
        mantisa = mantisa[:52]
    
    return mantisa
```

```python
def formato_IEEE_754(numero_decimal, exponente_notacion, notacion_cientifica, bits):
    """
    Genera la representación en formato IEEE 754 de un número decimal.

    Parámetros:
    numero_decimal (float): El número decimal que se desea convertir.
    exponente_notacion (int): El exponente obtenido de la notación científica binaria.
    notacion_cientifica (str): La mantisa en notación científica binaria (por ejemplo, '1.0001100110011').
    bits (int): Número de bits del formato IEEE 754 (32 para simple precisión, 64 para doble precisión).

    Descripción:
    - Calcula el bit de signo usando la función signo().
    - Calcula el exponente ajustado y su representación binaria usando la función exponente().
    - Calcula la mantisa ajustada usando la función mantisa().
    - Une los componentes en una cadena representando el formato IEEE 754.

    Retorna:
    str: Cadena con la representación del número en formato IEEE 754, separando signo, exponente y mantisa.
    """
    return f"{signo(numero_decimal)}  {exponente(exponente_notacion, bits)}  {mantisa(notacion_cientifica, bits)}"
```

```python
# Conversión de un número decimal a formato IEEE 754
# Número decimal a convertir
numero_decimal = 263.3

# 1ro Conversión a binario	
numero_binario_32 = decimal_a_binario(numero_decimal, 32)
print("El número decimal {} a binario es {}".format(numero_decimal, numero_binario_32)) 

# 2do Conversión a notación científica
notacion_cientifica_32, exponente_notacion_32 = notacion_cientifica(numero_binario_32)
print("El numero en binario {} a su forma en notacion cientifica {} x 2^{}".format(numero_binario_32, notacion_cientifica_32, exponente_notacion_32)) 

# 3ro Conversión a formato IEEE 754
print("El numero en formato IEEE 754 es: {}".format(formato_IEEE_754(numero_decimal, exponente_notacion_32, notacion_cientifica_32, 32)))
```

## Aritmetica de digitos finitos

Es la forma en que una computadora realiza operaciones matemátocas usando un número limitado de dígitos.

$± 0,d_1 d_2 d_3 d_4 d_5 ... d_k \cdot 10^n, 1 \leq  d1 \leq  9, n \text{ es entero}$

### Operaciones

1. $x ⨁ y = f_1(f(x) + f_1(y))$

```python
def representacion_notacion(signo, valor, exponente):
    """
    Genera una representación en notación científica de un número, adecuada para aritmética de dígitos finitos.

    Parámetros:
    signo (str): El signo del número, puede ser '-' para números negativos o '' para positivos.
    valor (float): El valor normalizado (mantisa) entre 0.1 y 1, obtenido tras ajustar el número original.
    exponente (int): El exponente de la potencia de 10 que acompaña a la mantisa.

    Descripción:
    - Toma el signo, la mantisa y el exponente, y los formatea en una cadena de texto en notación científica.
    - El formato de salida es: "<signo><mantisa> * 10^<exponente>".

    Retorna:
    str: Cadena con la representación del número en notación científica.
    """
    return f"{signo}{valor:.10g} * 10^{exponente}"
```

```python
def representacion_digitos_computador(numero):
    """
    Convierte un número decimal a su representación en notación científica normalizada para aritmética de dígitos finitos.

    Parámetros:
    numero (float): El número decimal que se desea representar en notación científica.

    Descripción:
    - Si el número es cero, retorna la representación estándar de cero en notación científica.
    - Para números distintos de cero, normaliza el valor para que esté en el rango [0.1, 1), ajustando el exponente en consecuencia.
    - Determina el signo del número.

    Retorna:
    tuple: Una tupla (signo, valor, exponente), donde:
        signo (str): '-' si el número es negativo, '' si es positivo.
        valor (float): La mantisa normalizada entre 0.1 y 1.
        exponente (int): El exponente de la potencia de 10.
    """
    if numero == 0:
        return "0.0 * 10^0"
    exponente = 0
    valor = abs(numero)
    while valor >= 1:
        valor /= 10
        exponente += 1
    while valor < 0.1:
        valor *= 10
        exponente -= 1
    signo = "-" if numero < 0 else ""
    return signo, valor, exponente
```

## Convergencia, Divergencia y Tolerancia

**Converge** si, al aumentar el número de interaciones, los resultados se acercan a la solución exacta.

```python
# Defnición de la función
def f(x):
    """
    Calcula el valor de la función f(x) = cos(x) - x.

    Parámetros:
    x (float): Valor en el que se evalúa la función.

    Retorna:
    float: Resultado de cos(x) - x.
    """
    return math.cos(x) - x

# Definición de la derivada de la función
def df(x):
    """
    Calcula la derivada de la función f(x) = cos(x) - x, es decir, f'(x) = -sin(x) - 1.

    Parámetros:
    x (float): Valor en el que se evalúa la derivada.

    Retorna:
    float: Resultado de -sin(x) - 1.
    """
    return -math.sin(x) - 1
```

```python
# Método de Newton-Raphson modificado para obtener todas las aproximaciones
def newton_raphson_modificado(xn_1, tol, iter):
    """
    Aplica el método de Newton-Raphson para encontrar la raíz de una función y almacena todas las aproximaciones intermedias.

    Parámetros:
    xn_1 (float): Valor inicial o aproximación inicial para la raíz.
    tol (float): Tolerancia para el criterio de parada.
    iter (int): Número máximo de iteraciones permitidas.

    Descripción:
    - Calcula aproximaciones a la raíz usando la fórmula de Newton-Raphson.
    - Guarda cada aproximación en una lista para graficar la convergencia.
    - El proceso se detiene si la diferencia entre dos aproximaciones consecutivas es menor que la tolerancia o si se alcanza el número máximo de iteraciones.

    Retorna:
    list: Lista con todas las aproximaciones generadas en cada iteración, incluyendo la inicial.
    """
    i = 1
    aproximaciones = [xn_1]
    while i <= iter:
        x_n = xn_1 - f(xn_1)/df(xn_1)
        aproximaciones.append(x_n)
        if abs(x_n - xn_1) < tol:
            return aproximaciones
        i += 1
        xn_1 = x_n
    return aproximaciones

# Implementacion del método de Newton-Raphson modificado
xn_1 = math.pi/4  # Aproximación inicial
tol = 1e-6      # Tolerancia
iteraciones = 10        # Número máximo de iteraciones

aproximaciones = newton_raphson_modificado(xn_1, tol, iteraciones)
```

```python
def graficar_convergencia(aprox, valor_exacto):
    """
    Grafica la convergencia de una secuencia de aproximaciones hacia un valor exacto.

    Parámetros:
    aprox (list): Lista de valores aproximados obtenidos en cada iteración del método numérico.
    valor_exacto (float): Valor exacto o referencia al que deberían converger las aproximaciones.

    Descripción:
    - Dibuja los valores de las aproximaciones en función del número de iteración.
    - Dibuja una línea horizontal indicando el valor exacto.
    - Añade etiquetas, leyenda, título y cuadrícula para mejor visualización.

    """
    plt.plot(range(len(aprox)), aprox, marker='o', label='Aproximación')
    plt.axhline(valor_exacto, color='r', linestyle='--', label='Valor exacto')
    plt.xlabel('Iteración')
    plt.ylabel('Valor aproximado')
    plt.title('Convergencia del método de Newton-Raphson para cos(x) = x')
    plt.legend()
    plt.grid(True)
    plt.show()

# Grafico de la convergencia:
valor_exacto = 0.7390851332  # Aproximación de la raíz de cos(x) = x
graficar_convergencia(aproximaciones, valor_exacto)
```

**Diverge** si los resultados se alejan de la solución exacta, o no se estabilizan.

```python
def secuencia_divergente(iteraciones):
    """
    Genera una secuencia numérica que diverge exponencialmente.

    Parámetros:
    iteraciones (int): Número de iteraciones o pasos a calcular en la secuencia.

    Descripción:
    - Comienza con un valor inicial x = 1.0.
    - En cada iteración, multiplica el valor anterior por 2 (x_{n+1} = 2 * x_n).
    - Almacena cada valor generado en una lista.

    Retorna:
    list: Lista de valores de la secuencia, incluyendo el valor inicial y los generados en cada iteración.
    """
    x = 1.0
    valores = [x]   
    for _ in range(iteraciones):
        x = 2 * x
        valores.append(x)
    return valores

iteraciones = 10
aproximaciones = secuencia_divergente(iteraciones)

graficar_convergencia(aproximaciones, 0)
```

**Tolerancia** es el límite aceptable de error que se impone para detener el método cuando se alcanza una solución lo "suficientemente buena".

```python
# Simulación de una secuencia que converge a 'resultado' con tolerancia
def simular_convergencia(resultado, r, tolerancia, iteraciones):
    """
    Simula una secuencia numérica que converge hacia un valor objetivo ('resultado').

    Parámetros:
    resultado (float): Valor al que se desea que la secuencia converja (valor exacto).
    r (float): Valor inicial de la secuencia.
    tolerancia (float): Límite aceptable de error para detener el programa.
    iteraciones (int): Número máximo de iteraciones permitidas.

    Descripción:
    - En cada iteración, la secuencia se acerca a 'resultado' usando una actualización recursiva.
    - Se almacena cada valor aproximado en una lista.
    - El proceso se detiene si la diferencia absoluta entre el valor actual y 'resultado' es menor que la tolerancia, o si se alcanza el número máximo de iteraciones.

    Retorna:
    list: Lista de valores aproximados generados en cada iteración, mostrando la convergencia hacia 'resultado'.
    """
    valores = []
    x = r
    for i in range(iteraciones):
        x = resultado - (resultado - x) * 0.5  # Aproxima a 'resultado'
        valores.append(x)
        if abs(x - resultado) < tolerancia:
            break
    return valores

# Parámetros de la simulación
resultado = 2.0         # Valor al que converge la secuencia (resultado exacto)
r = 0.5                 # Valor inicial
tolerancia = 0.08       # Tolerancia aceptable
iteraciones = 20        # Número máximo de iteraciones

valores = simular_convergencia(resultado, r, tolerancia, iteraciones)

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(valores) + 1), valores, marker='o', label='Aproximaciones')
plt.axhline(resultado, color='r', linestyle='--', label='Resultado exacto')
plt.axhline(resultado + tolerancia, color='g', linestyle=':', label='Límite superior tolerancia')
plt.axhline(resultado - tolerancia, color='g', linestyle=':', label='Límite inferior tolerancia')
plt.xlabel('Iteración')
plt.ylabel('Valor aproximado')
plt.title('Ejemplo de tolerancia en la convergencia')
plt.legend()
plt.grid(True)
plt.show()
```

## Criterios de parada

Tienen la funcion de evitar los ciclos infinitos.

### Error absoluto

Se define como:

$|P_N - P_{N-1}| < ∈$


```python
def criterio_parada_error_absoluto(aproximaciones, tolerancia):
    """
    Aplica el criterio de parada usando el error absoluto entre aproximaciones consecutivas.
    Imprime la aproximación y el error absoluto en cada iteración, y detiene si el error es menor que la tolerancia.

    Parámetros:
    aproximaciones (list): Lista de valores aproximados.
    tolerancia (float): Límite aceptable de error absoluto para detener el proceso.
    """
    for i in range(1, len(aproximaciones)):
        error_absoluto = abs(aproximaciones[i] - aproximaciones[i-1])
        print(f"Iteración {i}: Aproximación = {aproximaciones[i]}, Error absoluto = {error_absoluto}")
        if error_absoluto < tolerancia:
            print(f"Se cumple el criterio de parada en la iteración {i} (error absoluto < tolerancia)")
            break

# Ejemplo de uso del criterio de parada
tolerancia = 0.001  
aproximaciones = [1.0, 1.01, 1.005, 1.0025, 1.00125, 1.000625] 
criterio_parada_error_absoluto(aproximaciones, tolerancia)
```

### Error relativo

Se define como:

$\frac{|P_N - P_{N-1}|}{|P_N|} < ∈$, $P_N ≠ 0$

```python
def criterio_parada_error_relativo(aproximaciones, tolerancia):
    """
    Aplica el criterio de parada usando el error relativo entre aproximaciones consecutivas.
    Imprime la aproximación y el error relativo en cada iteración, y detiene si el error es menor que la tolerancia.

    Parámetros:
    aproximaciones (list): Lista de valores aproximados.
    tolerancia (float): Límite aceptable de error absoluto para detener el proceso.
    """
    for i in range(1, len(aproximaciones)):
        error_absoluto = abs((aproximaciones[i] - aproximaciones[i-1]) / aproximaciones[i])
        print(f"Iteración {i}: Aproximación = {aproximaciones[i]}, Error absoluto = {error_absoluto}")
        if error_absoluto < tolerancia:
            print(f"Se cumple el criterio de parada en la iteración {i} (error absoluto < tolerancia)")
            break

# Ejemplo de uso del criterio de parada
criterio_parada_error_relativo(aproximaciones, tolerancia)
```

### Valor de la función

Se define como:

$|f(P_N)| < ∈$

```python
def criterio_parada_valor_funcion(valor_x, funcion, funcion_derivada, tolerancia):
    """
    La función aplica el criterio de parada usando el valor absoluto de la función evaluada en cada aproximación.

    Parámetros:
    valor_x (float): Valor inicial o aproximación inicial para la raíz.
    funcion: Función a evaluar en cada aproximación.
    funcion_derivada: Derivada de la función.
    tolerancia (float): Límite aceptable para detener el proceso.

    Descripción:
    - Utiliza el método de Newton-Raphson para encontrar la raíz de la función.
    - En cada iteración, evalúa el valor absoluto de la función en la aproximación actual.
    - Imprime la aproximación y el valor absoluto de la función.
    - El proceso se detiene si el valor absoluto de la función es menor que la tolerancia.
    """
    iter = 0
    while True:
        valor_funcion = abs(funcion(valor_x))
        print(f"Iteración {iter}: Aproximación = {valor_x}, |f(x)| = {valor_funcion}")
        if valor_funcion < tolerancia:
            print(f"Se cumple el criterio de parada en la iteración {iter} (|f(x)| < tolerancia)")
            break
        iter += 1
        valor_x = valor_x - funcion(valor_x) / funcion_derivada(valor_x)

# Ejemplo de uso:
tolerancia = 0.001
x = math.pi / 4  # Aproximación inicial
criterio_parada_valor_funcion(x, f, df, tolerancia) # Se pasa como parámetro la función f, f(x) = cos(x) - x, y su derivada definida anteriormente.
```

### Numero de iteraciones

```python
def criterio_parada_numero_iteraciones(max_iter, x=30):
    """
    La función aplica el criterio de parada basado en el número máximo de iteraciones, se evalúa una función cualquiera, f(x) = x/2.

    Parámetros:
    max_iter (int): Número máximo de iteraciones permitidas.
    x (int): Valor inicial para la función, por defecto es 30.

    Imprime el número de cada iteración hasta alcanzar el máximo.
    """
    for i in range(1, max_iter + 1):
        print(f"Iteración {i}, valor de f(x) = {x/2}")
        x /= 2
    print(f"Se cumplió el criterio de parada: número de iteraciones, alcanzó {max_iter} iteraciones")

# Ejemplo de uso:
criterio_parada_numero_iteraciones(5)
```

 Tiempo de ejecucion se traduce a mayor cantidad de recursos del computador.

```python
def criterio_parada_tiempo_ejecucion(tiempo_max, x=30):
    """
    La función aplica el criterio de parada basado en el número máximo de iteraciones, se evalúa una función cualquiera, f(x) = x/2.

    Parámetros:
    tiempo_max (int): Número máximo de iteraciones permitidas.
    x (int): Valor inicial para la función, por defecto es 1s.

    Imprime el número de cada iteración hasta alcanzar el máximo.
    """
    tiempo_inicio = time.time()
    iter = 0
    while True:
        tiempo_actual = (time.time() - tiempo_inicio)
        print(f"Iteracion {iter}, valor de x = {x}, tiempo transcurrido: {tiempo_actual} s")
        if tiempo_actual > tiempo_max:
            print(f"Se cumple el criterio de parada: tiempo de ejecución > {tiempo_max} s")
            break
        x /= 2
        iter += 1
        time.sleep(0.09)

# Ejemplo de uso:
criterio_parada_tiempo_ejecucion(1)
```

# Método de la bisección

Realiza divisiones sucesivas infinitas hasta aproximarse a un valor real, el error se va disminuyendo y acercandose a cero. 

```python
# Función de ejemplo
def func(x):
    """
    Calcula el valor de la función f(x) = x^2 - 1 - e^{1 - x^2}.

    Parámetros:
    x (float): Valor x en el que se evalúa la función.

    Retorna:
    float: Resultado de x^2 - 1 - exp(1 - x^2).
    """
    return x**2 - 1 - math.exp(1 - x**2)
```

```python
# Función para calcular el signo de un valor
def signum(value):
    """
    Calcula el signo de un valor numérico.

    Parámetros:
    value (float o int): El valor numérico al que se le desea calcular el signo.

    Retorna:
    int: 1 si el valor es positivo, -1 si es negativo, 0 si es cero.
    """
    if value > 0:
        return 1
    elif value < 0:
        return -1
    else:
        return 0
```

```python
# Método de la bisección
def biseccion(func, a, b, tolerance, max_iter=1000):
    """
    La funcion implementa el método de la bisección para encontrar una raíz de la función 'func' en el intervalo [a, b].

    Parámetros:
    func: Función para la cual se busca la raíz.
    a (float): Extremo izquierdo del intervalo.
    b (float): Extremo derecho del intervalo.
    tolerance (float): Tolerancia para el criterio de parada.
    max_iter (int, opcional): Número máximo de iteraciones permitidas (por defecto 1000).

    Descripción:
    - Calcula el punto medio p del intervalo continuo [a, b].
    - Evalúa la función en p (FP) y en a (FA).
    - Imprime los valores de cada iteración.
    - Si FP es cero o el intervalo es menor que la tolerancia, retorna la solución aproximada.
    - Si el signo de FA y FP es igual, actualiza a = p; si no, actualiza b = p.
    - Si no se encuentra la raíz en max_iter iteraciones, informa el fracaso.

    Retorna:
    float o None: La raíz aproximada si se encuentra dentro de la tolerancia, None si no converge.
    """
    i = 1
    FA = func(a)

    while i <= max_iter:
        p = a + (b - a) / 2
        FP = func(p)

        print(f"Iteración {i}: a = {a}, b = {b}, p = {p}, FA = {FA}, FP = {FP}")

        if FP == 0 or (b - a) / 2 < tolerance:
            print("Procedimiento completado exitosamente. \n Solución aproximada p =", p, "en la iteración", i)
            return p
        i += 1

        if signum(FA) * signum(FP) > 0:  # Usa signum para determinar el cambio de signo
            a = p
            FA = FP
        else:
            b = p

    print(f"El método fracasó despues de {max_iter} iteraciones.")
    return None    
```

```python
# Aplicación del método de bisección
a, b = -2, 0
tolerance = 10**(-3)
biseccion(func, a, b, tolerance)
```

**Funciones válidas para el método de la bisección**

El método de la bisección puede aplicarse a una amplia variedad de funciones, siempre que sean continuas en el intervalo $[a, b]$ y cumplan con el requisito de cambio de signo: $f(a) \cdot f(b) < 0$.

| Tipo de función      | Ejemplo de función (`func`)                | Observaciones principales                                  |
|----------------------|--------------------------------------------|-----------------------------------------------------------|
| Polinómicas          | `x**2 - 4`                       | Raíces reales y complejas, debe haber cambio de signo     |
| Trigonométricas      | ` math.cos(x) - x`                | Puede tener múltiples raíces, elegir bien el intervalo    |
| Exponenciales        | ` math.exp(x) - 2`                | Generalmente continuas, verificar extremos                |
| Logarítmicas         | ` math.log(x) - 1`                | Dominio restringido, evitar extremos no definidos         |
| Racionales           | ` (x-1)/(x+2)`                    | Evitar discontinuidades dentro del intervalo              |
| Continuas a trozos   | ` x if x < 0 else x**2 - 1`       | Debe ser continua en \([a, b]\)                           |

**Observaciones importantes del método:**
- La función debe ser continua en el intervalo $[a, b]$.
- Es obligatorio que $f(a)$ y $f(b)$ tengan signos opuestos (exista cambio de signo).
- El intervalo debe ser correctamente elegido para contener la raíz.
- No garantiza encontrar raíces dobles o de orden par si no hay cambio de signo.
- Puede ser lento si la raíz está cerca de uno de los extremos.


# Método de Newton-Raphson

El método de Newton-Raphson es un método iterativo para encontrar raíces de funciones. Se basa en la idea de aproximar la función por su tangente en un punto inicial y luego encontrar el punto donde esa tangente cruza el eje x.

```python
# Método de Newton-Raphson
def newton_raphson(xn_1, tol, iter, fun, dfun):
    """
    La función aplica el método de Newton-Raphson para encontrar una raíz de la función f(x).

    Parámetros:
    xn_1 (float): Valor inicial o aproximación inicial para la raíz.
    tol (float): Tolerancia para el criterio de parada.
    iter (int): Número máximo de iteraciones permitidas.
    fun: Función para la cual se busca la raíz.
    dfun: Derivada de la función para aplicar el método.

    Descripción:
    - Calcula aproximaciones sucesivas a la raíz usando la fórmula de Newton-Raphson.
    - Imprime los valores de cada iteración.
    - El proceso se detiene si la diferencia entre dos aproximaciones consecutivas es menor que la tolerancia o si se alcanza el número máximo de iteraciones.

    Retorna:
    None. Imprime la raíz aproximada y el número de iteraciones si el método converge, o un mensaje de fallo si no converge.
    """
    i = 1
    while i <= iter:
        x_n = xn_1 - fun(xn_1)/dfun(xn_1)

        print(f"Iteración {i}: xn_1 = {xn_1}, x_n = {x_n}, f(xn_1) = {fun(xn_1)}, f'(x_n) = {dfun(xn_1)}")

        if abs(x_n - xn_1) < tol:
            print(f"El procedimiento fue exitoso. \n Raíz aproximada: {x_n} \n Iteraciones: {i}")
            return
        i += 1
        xn_1 = x_n
    print(f"El método falló después de {iter} iteraciones.")
    return None
```

```python
# Ejemplo de uso con la función coseno analizada anteriormente:
xn_1 = math.pi/4  # Aproximación inicial
tol = 1e-6      # Tolerancia
iter = 100        # Número máximo de iteraciones

newton_raphson(xn_1, tol, iter, f, df)
```

**Funciones válidas para el método de Newton-Raphson**

El método de Newton-Raphson puede aplicarse a una amplia variedad de funciones, siempre que cumplan:

| Tipo de función      | Ejemplo de función (`f(x)`)                | Observaciones principales                                  |
|----------------------|--------------------------------------------|-----------------------------------------------------------|
| Polinómicas          | `x**2 - 4`                                 | Rápida convergencia si la derivada no es cero en la raíz  |
| Trigonométricas      | `math.cos(x) - x`                          | Puede tener múltiples raíces, elegir bien el punto inicial|
| Exponenciales        | `math.exp(x) - 2`                          | Derivada nunca es cero, buena convergencia                |
| Logarítmicas         | `math.log(x) - 1`                          | Dominio restringido, evitar puntos donde la derivada diverge|
| Racionales           | `(x-1)/(x+2)`                              | Evitar raíces donde la derivada sea cero o indefinida     |
| Combinadas           | `x**3 - math.cos(x)`                       | Asegurar derivabilidad y buen punto inicial               |

**Observaciones importantes del método:**
- La función debe ser derivable en el intervalo de interés.
- La derivada no debe ser cero cerca de la raíz (evitar divisiones por cero).
- La elección del punto inicial es crucial para la convergencia.
- Puede no converger si la función es muy plana o la derivada cambia de signo cerca de la raíz.
- Puede converger rápidamente si se cumplen las condiciones adecuadas.

## Método de la secante

Tiene como objetivo encontrar raices. Ya no tiene derivadas. 

```python
def secante(xn_2, xn_1, tol, iter, fun):
    """
    La función aplica el método de la secante para encontrar una raíz de la función f(x).

    Parámetros:
    xn_2 (float): Primer valor inicial para la secuencia (x_{n-2}).
    xn_1 (float): Segundo valor inicial para la secuencia (x_{n-1}).
    tol (float): Tolerancia para el criterio de parada.
    iter (int): Número máximo de iteraciones permitidas.
    fun: Función para la cual se busca la raíz.

    Descripción:
    - Calcula aproximaciones sucesivas a la raíz usando la fórmula de la secante.
    - Imprime los valores de cada iteración.
    - El proceso se detiene si la diferencia entre dos aproximaciones consecutivas es menor que la tolerancia o si se alcanza el número máximo de iteraciones.
    - Si el denominador es cero, informa que el método falla por división por cero.

    Retorna:
    None. Imprime la raíz aproximada y el número de iteraciones si el método converge, o un mensaje de fallo si no converge.
    """
    i = 2
    while i <= iter:
        if fun(xn_1) - f(xn_2) == 0:
            print("División por cero detectada. El método falla.")
            return None
        x_n = xn_1 - fun(xn_1) * (xn_1 - xn_2) / (fun(xn_1) - f(xn_2))

        print(f"Iteración {i-1}: xn_2 = {xn_2}, xn_1 = {xn_1}, x_n = {x_n}, f(xn_2) = {fun(xn_2)}, f(xn_1) = {fun(xn_1)}")

        if abs(x_n - xn_1) < tol:
            print(f"El procedimiento fue exitoso. \n Raíz aproximada: {x_n} \n Iteraciones: {i-1}")
            return
        xn_2, xn_1 = xn_1, x_n
        i += 1
    print(f"El método falló después de {iter} iteraciones.")
    return None
```

```python
# Ejemplo de uso con la función coseno analizada anteriormente en el método de Newton-Raphson:
xn_2 = 0.5      # Primer valor inicial
xn_1 = math.pi/4      # Segundo valor inicial
tol = 1e-6    # Tolerancia
iter = 100      # Número máximo de iteraciones

secante(xn_2, xn_1, tol, iter, f)
```

**Funciones válidas para el método de la secante**

| Tipo de función      | Ejemplo de función (`f(x)`)                | Observaciones principales                                  |
|----------------------|--------------------------------------------|-----------------------------------------------------------|
| Polinómicas          | `x**2 - 4`                                 | Puede converger rápido, evitar valores iniciales con mismo valor de función |
| Trigonométricas      | `math.cos(x) - x`                          | Puede tener múltiples raíces, elegir bien los puntos iniciales |
| Exponenciales        | `math.exp(x) - 2`                          | Generalmente buena convergencia, evitar valores iniciales iguales |
| Logarítmicas         | `math.log(x) - 1`                          | Dominio restringido, evitar extremos no definidos         |
| Racionales           | `(x-1)/(x+2)`                              | Evitar discontinuidades y denominadores cero              |
| Combinadas           | `x**3 - math.cos(x)`                       | Asegurar continuidad y buenos puntos iniciales            |

**Observaciones importantes del método:**
- La función debe ser continua en el intervalo de interés.
- Los valores iniciales deben ser distintos y producir valores de función diferentes (evitar división por cero).
- No requiere derivada, pero puede fallar si el denominador es cero.
- La elección de los puntos iniciales afecta la convergencia.
- Puede no converger si la función es muy plana o los puntos iniciales están mal elegidos.

# **Método de la aproximación falsa**

Maneja las mismas condiciones que el metodo de la secante pero agrega una condición adicional que permite que se llegue a la solucion. 

```python
def posicion_falsa(xn_2, xn_1, tol, iter, fun):
    """
    La función aplica el método de la aproximación falsa (Regula Falsi) para encontrar una raíz de la función f(x).

    Parámetros:
    xn_2 (float): Primer valor inicial del intervalo (extremo izquierdo).
    xn_1 (float): Segundo valor inicial del intervalo (extremo derecho).
    tol (float): Tolerancia para el criterio de parada.
    iter (int): Número máximo de iteraciones permitidas.
    fun: Función para la cual se busca la raíz.

    Descripción:
    - Calcula aproximaciones sucesivas a la raíz usando la fórmula de la posicion falsa.
    - Imprime los valores de cada iteración.
    - El proceso se detiene si la diferencia entre dos aproximaciones consecutivas es menor que la tolerancia o si se alcanza el número máximo de iteraciones.
    - Si el denominador es cero, informa que el método falla por división por cero.
    - Actualiza los extremos del intervalo según el signo de la función evaluada.

    Retorna:
    None. Imprime la raíz aproximada y el número de iteraciones si el método converge, o un mensaje de fallo si no converge.
    """
    i = 2
    f_xn_2 = fun(xn_2)
    f_xn_1 = fun(xn_1)
    while i <= iter:
        if f_xn_1 - f_xn_2 == 0:
            print("División por cero detectada. El método falla.")
            return None
        
        x_n = xn_1 - f_xn_1 * (xn_1 - xn_2) / (f_xn_1 - f_xn_2)

        print(f"Iteración {i-1}: xn_2 = {xn_2}, xn_1 = {xn_1}, x_n = {x_n}, f(xn_2) = {fun(xn_2)}, f(x_n) = {fun(x_n)}")

        if abs(x_n - xn_1) < tol:
            print(f"El procedimiento fue exitoso. \n Raíz aproximada: {x_n} \n Iteraciones: {i-1}")
            return
        i += 1
        q = fun(x_n)
        if q * f_xn_1 < 0:
            xn_2 = xn_1
            f_xn_2 = f_xn_1
        xn_1 = x_n
        f_xn_1 = q
    print(f"El método falló después de {iter} iteraciones.")
    return None
```

# Series de Taylor

## **El polinomio de Taylor**

$f(x) = P_n(x) + R_n(x)$ 

donde $P_n(x)$ es el polinomio de Taylor y $R_n(x)$ es el residuo o error de la aproximación con **truncamiento**.

```python
def polinomio_taylor(func_str, x0, grado, decimales):
    
    """
    :func_str: string, función original
    :x0: entero, punto alrededor del cual se desarrolla el polinomio    
    :grado: entero, grado de polinomio de Taylor    
    :decimales: entero, decimales a truncar
    :return: polinomio de taylor aproximado a la función origianl, f(x) ≈ P_n(x) + R_n(x)
    """
    
    x = sp.Symbol('x')
    f = sp.sympify(func_str)

    taylor_expr = 0

    for n in range(grado + 1):
        derivada = f.diff(x, n)
        deriv_val = derivada.subs(x, x0)
        factorial_n = sp.factorial(n)
        coef = deriv_val / factorial_n

        if coef.is_number:
            coef = truncar(float(coef), decimales)

        if coef == 0:
            continue

        base = (x - x0)**n if x0 != 0 else x**n
        taylor_expr += coef * base

    # Primero se expande el polinomio 
    taylor_expr = sp.expand(taylor_expr)
    # Luego se extrae los coeficientes con la ayuda de Poly de sympy
    taylor_poly = sp.Poly(taylor_expr, x)
    coeficiente_list = taylor_poly.all_coeffs()
    grado_actual = len(coeficiente_list) - 1

    # Se truncan todos los coeficientes
    coeficientes_truncados = [truncar(float(c), decimales) for c in coeficiente_list]

    # Finalmente, se reconstreye el polinomio con los coeficientes truncados
    taylor_truncado = 0
    for i, coeficiente in enumerate(coeficientes_truncados):
        taylor_truncado += coeficiente*x**(grado_actual - i)
    
    taylor_truncado = sp.expand(taylor_truncado)
    taylor_truncado = sp.collect(taylor_truncado, x)
    Pn_str = str(taylor_truncado)

    n_plus_1 = grado + 1
    factorial_np1 = sp.factorial(n_plus_1)
    inverso_truncado = truncar(1 / float(factorial_np1), decimales)

    base_error = sp.expand((x - x0)**n_plus_1)
    Rn_str = f"( {inverso_truncado}*{func_str}^({n_plus_1})(ξ(x)) )*({base_error})"

    print(f"P_n(x) ≈ {Pn_str}")
    print(f"R_n(x) ≈ {Rn_str}")

    return f"f(x) ≈ {Pn_str} + {Rn_str}"
```

```python
print(polinomio_taylor("exp(sin(x))", x0=1, grado=3, decimales=4))
```

## **Series de Maclaurin**

- Es una serie de Taylor que funcionan cuando $x_0 = 0$

-  **No se usa en interpolación** porque mientras más se va aproximando más se va alejando. Por lo que solo se usa en aproximación.

- Son aproximaciones locales.

```python
def serie_maclaurin(func_str, grado, decimales):
    """
    Calcula el polinomio de Maclaurin (Taylor en x0=0) de una función dada como string.

    :param func_str: string, función original (por ejemplo, "sin(x)")
    :param grado: int, grado del polinomio de Maclaurin
    :param decimales: int, número de decimales a truncar
    :return: string, polinomio de Maclaurin aproximado a la función original
    """
    x = sp.Symbol('x')
    f = sp.sympify(func_str)
    maclaurin_expr = 0

    for n in range(grado + 1):
        derivada = f.diff(x, n)
        deriv_val = derivada.subs(x, 0)
        factorial_n = sp.factorial(n)
        coef = deriv_val / factorial_n

        if coef.is_number:
            coef = truncar(float(coef), decimales)

        if coef == 0:
            continue

        maclaurin_expr += coef * x**n

    maclaurin_expr = sp.expand(maclaurin_expr)
    maclaurin_poly = sp.Poly(maclaurin_expr, x)
    coeficiente_list = maclaurin_poly.all_coeffs()
    grado_actual = len(coeficiente_list) - 1

    coeficientes_truncados = [truncar(float(c), decimales) for c in coeficiente_list]

    maclaurin_truncado = 0
    for i, coeficiente in enumerate(coeficientes_truncados):
        maclaurin_truncado += coeficiente * x**(grado_actual - i)

    maclaurin_truncado = sp.expand(maclaurin_truncado)
    maclaurin_truncado = sp.collect(maclaurin_truncado, x)
    Pn_str = str(maclaurin_truncado)

    n_plus_1 = grado + 1
    factorial_np1 = sp.factorial(n_plus_1)
    inverso_truncado = truncar(1 / float(factorial_np1), decimales)
    base_error = sp.expand(x**n_plus_1)
    Rn_str = f"( {inverso_truncado}*{func_str}^({n_plus_1})(ξ(x)) )*({base_error})"

    print(f"P_n(x) ≈ {Pn_str}")
    print(f"R_n(x) ≈ {Rn_str}")

    return f"f(x) ≈ {Pn_str} + {Rn_str}"
```

```python
print(serie_maclaurin("exp(sin(x))", grado=3, decimales=4))
```

### Polinomios de Lagrange

No usa derivadas, solo usa puntos, o sea con los puntos se va armando el polinomio.

Se define como:

<div align="center">

$P(x) = \sum_{k=0}^{n} y_k \cdot L_k(x)$

</div>

donde

<div align="center">

$L_k(x) = \prod_{i=0, i \neq k}^{n} \frac{(x - x_i)}{(x_k - x_i)}$

</div>



```python
def polinomio_lagrange(x_valores, y_valores):
    """
    Calcula el polinomio de Lagrange para un conjunto de puntos dados.

    Parámetros:
    x_valores (list): Lista de valores x de los puntos.
    y_valores (list): Lista de valores y de los puntos.

    Retorna:
    sympy.Expr: Polinomio de Lagrange como expresión simbólica.
    """
    x = sp.Symbol('x')
    n = len(x_valores)
    Polinomio = 0
    for k in range(n):
        Lk = 1
        for i in range(n):
            if i != k:
                Lk *= (x - x_valores[i]) / (x_valores[k] - x_valores[i])
        Polinomio += y_valores[k] * Lk
    Polinomio = sp.expand(Polinomio)
    print(f"P(x) = {Polinomio}")
    return None
```

```python
# Ejemplo de uso:
x_valores = [2, 3.2, 4]
y_valores = [1.43, 2.79, 3.56]

polinomio_lagrange(x_valores, y_valores)
```

# Splines cúbicos

Los splines cúbicos son polinomios de tercer grado que se utilizan para interpolar puntos de datos. Se caracterizan por ser suaves y continuos, lo que los hace ideales para representar curvas.

## **Spline cubico con frontera natural**


```python
def spline_cubico_natural(x_valores, y_valores):
    """
    Calcula los coeficientes del spline cúbico natural para los puntos dados.

    Parámetros:
    x_valores (list or array): nodos x, de tamaño n+1
    y_valores (list or array): valores y = f(x), de tamaño n+1

    Retorna:
    a, b, c, d: listas de coeficientes para cada subintervalo [x_j, x_{j+1}]
    """
    n = len(x_valores) - 1
    a = y_valores[:]
    h = []

    # Paso 1: Calcular las diferencias divididas
    for i in range(n):
        h.append(x_valores[i+1] - x_valores[i])

    # Paso 2: Calcular alpha
    alpha = [0] * (n+1)
    for i in range(1, n):
        alpha[i] = (3/h[i]) * (a[i+1] - a[i]) - (3/h[i-1]) * (a[i] - a[i-1])

    # Paso 3
    l = [1] + [0]*n
    u = [0]*(n+1)
    z = [0]*(n+1)

    # Paso 4
    for i in range(1, n):
        l[i] = 2*(x_valores[i+1] - x_valores[i-1]) - h[i-1]*u[i-1]
        u[i] = h[i]/l[i]
        z[i] = (alpha[i] - h[i-1]*z[i-1])/l[i]

    l[n] = 1
    z[n] = 0
    c = [0]*(n+1)

    # Paso 6
    b = [0]*n
    d = [0]*n
    for j in range(n-1, -1, -1):
        c[j] = z[j] - u[j]*c[j+1]
        b[j] = (a[j+1] - a[j])/h[j] - h[j]*(c[j+1] + 2*c[j])/3
        d[j] = (c[j+1] - c[j])/(3*h[j])

    # Paso 7: Armar los splines
    for i in range(n):
        print(f"S_{i}(x) = {truncar(a[i], 5)} + ({truncar(b[i], 5)})(x - ({truncar(x_valores[i], 5)})) + ({truncar(c[i], 5)})(x - ({x_valores[i]}))^2 + ({truncar(d[i], 5)})(x - ({truncar(x_valores[i], 5)}))^3")

    return a[:-1], b, c[:-1], d
```

```python
# Ejemplo de uso:
x_valores = [-2, -1, 1, 3]
y_valores = [3, 1, 2, -1]

spline_cubico_natural(x_valores, y_valores)
```

## **Spline cúbico con frontera condicional**

```python
def spline_cubico_frontera_condicional(x, y, B0, Bn):
    """
    Calcula los coeficientes del spline cúbico con frontera condicional (derivadas dadas en los extremos).

    Parámetros:
    x: lista de valores x (n+1 elementos)
    y: lista de valores y (n+1 elementos)
    B0: derivada en el extremo izquierdo (x0)
    Bn: derivada en el extremo derecho (xn)

    Retorna:
    a, b, c, d: listas de coeficientes para cada subintervalo [x_j, x_{j+1}]
    """
    n = len(x) - 1
    a = y[:]
    h = []

    # Paso 1: Calcular las diferencias h
    for i in range(n):
        h.append(x[i+1] - x[i])

    #h = [x[i+1] - x[i] for i in range(n)]

    # Paso 2 y 3: alpha
    alpha = [0] * (n+1)
    alpha[0] = 3 * (a[1] - a[0]) / h[0] - 3 * B0
    alpha[n] = 3 * Bn - 3 * (a[n] - a[n-1]) / h[n-1]
    for i in range(1, n):
        alpha[i] = 3/h[i] * (a[i+1] - a[i]) - 3/h[i-1] * (a[i] - a[i-1])

    # Paso 4: l, u, z
    l = [0] * (n+1)
    u = [0] * (n+1)
    z = [0] * (n+1)
    l[0] = 2 * h[0]
    u[0] = 0.5
    z[0] = alpha[0] / l[0]

    # Paso 5
    for i in range(1, n):
        l[i] = 2 * (x[i+1] - x[i-1]) - h[i-1] * u[i-1]
        u[i] = h[i] / l[i]
        z[i] = (alpha[i] - h[i-1] * z[i-1]) / l[i]

    # Paso 6 y 7: c, b, d
    l[n] = h[n-1] * (2 - u[n-1])
    z[n] = (alpha[n] - h[n-1] * z[n-1]) / l[n]

    c = [0] * (n+1)
    b = [0] * n
    d = [0] * n
    c[n] = z[n]

    for j in range(n-1, -1, -1):
        c[j] = z[j] - u[j] * c[j+1]
        b[j] = (a[j+1] - a[j]) / h[j] - h[j] * (c[j+1] + 2 * c[j]) / 3
        d[j] = (c[j+1] - c[j]) / (3 * h[j])

    # Mostrar los polinomios
    for j in range(n):
        print(f"S_{j}(x) = ({truncar(a[j], 5)}) + ({truncar(b[j], 5)})*(x - ({truncar(x[j], 5)})) + ({truncar(c[j], 5)})*(x - ({truncar(x[j], 5)}))**2 + ({truncar(d[j], 5)})*(x - ({truncar(x[j], 5)}))**3")

    return a[:-1], b, c[:-1], d
```

```python
# Ejemplo de uso:
x_valores = [-2, -1, 1, 3]
y_valores = [3, 1, 2, -1]

spline_cubico_frontera_condicional(x_valores, y_valores, 1, -1)
```

# **Mínimos cuadrados**

Los **minimos cuadrados** son un método de aproximación que minimiza la suma de los cuadrados de los errores (residuos) entre los valores observados y los valores predichos por un modelo. Este método es ampliamente utilizado en regresión lineal y otros modelos estadísticos.


```python
def minimos_cuadrados(x, y):
    """
    Calcula la recta de regresión por mínimos cuadrados para los puntos dados.

    Parámetros:
    x (list o array): valores de x
    y (list o array): valores de y

    Retorna:
    (float, float): coeficientes (pendiente m, intercepto b) de la recta y = m*x + b
    """
    x = np.array(x)
    y = np.array(y)
    n = len(x)
    sum_x = np.sum(x)
    sum_y = np.sum(y)
    sum_x2 = np.sum(x**2)
    sum_xy = np.sum(x*y)

    m = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x**2)
    b = (sum_y - m * sum_x) / n

    print(f"Recta de mínimos cuadrados: y = {m:.4f}x + {b:.4f}")
    return m, b
```

# Eliminación Gaussiana

```python
def eliminacion_gaussiana(matriz_A):
    """
    Resuelve un sistema de ecuaciones lineales Ax = b usando el método de eliminación gaussiana.

    Parámetros:
    matriz_A (list): Matriz aumentada [A|b] donde A es la matriz de coeficientes nxn y b es el vector de términos independientes.

    Descripción:
    - Aplica eliminación hacia adelante para convertir la matriz en triangular superior.
    - Hace ceros debajo de la diagonal principal mediante operaciones elementales de fila.
    - Verifica que no haya elementos cero en la diagonal.
    - Utiliza sustitución hacia atrás para resolver el sistema triangular superior.
    - El proceso se detiene si se detecta que no existe solución única.

    Retorna:
    list: Vector solución x con los valores de las variables. 
    Imprime las soluciones con 6 decimales de precisión.
    """
    n = len(matriz_A)

    # Hacer ceros hacia abajo
    for col in range(n-1):
        for row in range(col+1, n):
            factor = matriz_A[row][col]/matriz_A[col][col]
            for j in range(col, n+1):
                matriz_A[row][j] -= factor * matriz_A[col][j]
        if matriz_A[col][col] == 0:
            print("No existe solucion unica")
            return
            
    x=[0]*n
    for i in range(n-1, -1, -1):
        suma = 0
        for j in range(i+1, n):
            suma += matriz_A[i][j] * x[j]
        x[i] = (matriz_A[i][n] - suma) / matriz_A[i][i]
        
    # Impresión de la solución
    for i in range(n):
        print(f"x{i+1} = {x[i]:.6f}")
    
    return x

```

```python
matriz_aumentada = [    [3, 2, 1, 1, 1],
    [2, 3, 3, 2, 4],
    [5, 4, 6, 3, 7],
    [4, 3, 5, 6, 10] ]

print(eliminacion_gaussiana(matriz_aumentada))
```

# Gauss-Jordan

```python
def gauss_jordan(matriz_A):
    """
    Resuelve un sistema de ecuaciones lineales Ax = b usando el método de Gauss-Jordan.

    Parámetros:
    matriz_A (list): Matriz aumentada [A|b] donde A es la matriz de coeficientes nxn y b es el vector de términos independientes.

    Descripción:
    - Aplica eliminación hacia adelante para hacer ceros debajo de la diagonal principal.
    - Aplica eliminación hacia atrás para hacer ceros arriba de la diagonal principal.
    - Normaliza la diagonal principal para convertirla en unos (forma escalonada reducida).
    - Verifica que no haya elementos cero en la diagonal (singularidad).
    - Obtiene directamente las soluciones de la última columna de la matriz reducida.
    - El proceso se detiene si se detecta que no existe solución única.

    Retorna:
    None. Imprime las soluciones x1, x2, ..., xn con 6 decimales de precisión.
    """

    n = len(matriz_A)

    # Hacer ceros hacia abajo
    for col in range(n-1):
        for row in range(col+1, n):
            factor = matriz_A[row][col]/matriz_A[col][col]
            for j in range(col, n+1):
                matriz_A[row][j] -= factor * matriz_A[col][j]
        if matriz_A[col][col] == 0:
            print("No existe solucion unica")
            return    
        
    # Hacer ceros hacia arriba
    for col in range(n-1, 0, -1):
        for row in range(col-1, -1, -1):
            factor = matriz_A[row][col]/ matriz_A[col][col]
            for j in range(n+1):
                matriz_A[row][j] -= factor * matriz_A[col][j]

    # Hacer 1 en la diagonal
    for i in range(n):
        valor = matriz_A[i][i]
        for j in range(n+1):
            matriz_A[i][j] /= valor

    # Soluciones
    for i in range(n):
        print(f"x{i+1} = {matriz_A[i][n]:.6f}")
```

# **Descomposición LU**

```python
def determinante(matriz):
    n = len(matriz)
    deter = 1
    A = [row[:] for row in matriz]
    
    # Verificar que la matriz sea cuadrada
    for row in matriz:
        if len(row) != n:
            print("No es una matriz cuadrada")
            return None

    if n == 1:
        return matriz[0][0]
    if n == 2:
        return matriz[0][0] * matriz[1][1] - matriz[0][1] * matriz[1][0]
    for k in range(n):
        max_row = k
        for i in range(k+1, n):
            if abs(A[i][k]) > abs(A[max_row][k]):
                max_row = i 
        if A[max_row][k] == 0:
            return 0
        if max_row != k:
            A[k], A[max_row] = A[max_row], A[k]
            deter = -deter
        deter *= A[k][k]
        for row in range(k+1, n):
            factor = A[row][k] / A[k][k]
            for col in range(k, n):
                A[row][col] -= factor * A[k][col]

    if deter == 0:
        print("El sistema no tiene solucion unica, el determinante es 0")
    else:
        print("El sistema tiene solucion unica y el determinante es:", round(deter, 2))
```

## Método de la Matriz Inversa

```python
def metodo_matriz_inversa(A, b):
    """
    A: matriz de coeficientes (lista de listas)
    b: vector de términos independientes (lista)
    Resuelve Ax = b calculando x = A⁻¹ * b
    """
    n = len(A)
    
    # Matriz indentidad
    identidad = []
    for i in range(n):
        fila = [0]*n
        fila[i] = 1
        identidad.append(fila)

    # Matriz aumentada [A | I]
    matriz_aumentada = []
    for i in range(n):
        fila = A[i] + identidad[i]
        matriz_aumentada.append(fila)

    # Calcular la matriz inversa
    for col in range(n):
        for row in range(col+1, n):
            factor = matriz_aumentada[row][col] / matriz_aumentada[col][col]
            for j in range(2*n):
                matriz_aumentada[row][j] -= factor * matriz_aumentada[col][j]
    
    for col in range(n-1, -1, -1):
        for row in range(col-1, -1, -1):
            factor = matriz_aumentada[row][col] / matriz_aumentada[col][col]
            for j in range(2*n):
                matriz_aumentada[row][j] -= factor * matriz_aumentada[col][j]
    
    for i in range(n):
        valor = matriz_aumentada[i][i]
        for j in range(2*n):
            matriz_aumentada[i][j] /= valor

    # Extraer la matriz inversa
    A_inversa = []
    for i in range(n):
        fila_inversa = []
        for j in range(n, 2*n):
            fila_inversa.append(matriz_aumentada[i][j])
        A_inversa.append(fila_inversa)

    # Imprimir la matriz inversa
    print("Matriz Inversa A⁻¹:")
    for row in range(n):
        for col in range(n):
            print(f"{A_inversa[row][col]:.6f}", end=" ")
        print()

    # Multiplicar A⁻¹ * b para obtener los valores de x
    x_valores = []
    for i in range(n):
        suma = 0
        for j in range(n):
            suma += A_inversa[i][j] * b[j]
        x_valores.append(suma)

    # Soluciones de x
    print("\n Soluciones:")
    for i in range(n):
        print(f"x{i+1} = {x_valores[i]:.6f}")
```

```python
A = [[1, 2, 3], [0, 1, 4], [5, 6, 0]]
b = [1, 2, 3]  
print("Verificar el determinante de la matriz A")
determinante(A)
print("\n")
metodo_matriz_inversa(A, b)
```

## Producto de Matrices, Doolittle

```python
def descomposicion_LU_doolittle(A):
    """
    La funcion descompone una matriz A en el producto L × U, usando el metodo de Doolittle.
    A: matriz cuadrada 
    Retorna: L (triangular inferior), U (triangular superior)
    """
    n = len(A)
    L,  U= [], []

    # Verificar que sea matriz cuadrada
    for fila in A:
        if len(fila) != n:
            print("No es una matriz cuadrada")
            return None, None
    
    # Llenar matrices LU
    for i in range(n):
        fila_L = []
        fila_U = []
        for j in range(n):
            if i == j:
                fila_L.append(1)
            else:
                fila_L.append(0)
            fila_U.append(0)
        L.append(fila_L)
        U.append(fila_U)

    # Calculo de LU
    for i in range(n):
        for k in range(i, n):
            suma = 0
            for j in range(i):
                suma += L[i][j] * U[j][k]
            U[i][k] = A[i][k] - suma
    
        for k in range(i+1, n):
            suma = 0
            for j in range(i):
                suma += L[k][j] * U[j][i]
            L[k][i] = (A[k][i] - suma) / U[i][i]
    
    # Impresion de L
    print("L = ")
    for row in range(n):
        for col in range(n):
            print(f"{L[row][col]:.6f}", end=" ")
        print("")

    # Impresion de U
    print("\n U = ")
    for row in range(n):
        for col in range(n):
            print(f"{U[row][col]:.6f}", end=" ")
        print("")
    
    return L, U
```
